```python
from Bio import SeqIO
from pathlib import Path

directory = "/kaggle/input/100-genes"

for filename in Path(directory).glob("*.fasta"):
    sequences = []

    for sequence in SeqIO.parse(filename, "fasta"):
        sequences.append(sequence.seq) #keeping as biopython seq object only
        
    genes[f"{str(filename).split('/')[-1].rstrip('.fasta')}"] = sequences
```

```python
from Bio import Align
aligner = Align.PairwiseAligner()

def align(seq1, seq2):
    alignments = aligner.align(seq1, seq2)
    return alignments.score

def hammingAlign(seq1, seq2):
    seq1, seq2 = list(seq1), list(seq2)
    return hamming(seq1, seq2)
```
